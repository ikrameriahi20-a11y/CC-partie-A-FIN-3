Régression linéairePrésentationNom et famille : Régression linéaire / Linear Regression, appartenant à la famille de la régression supervisée.Objectif : Ajuster une droite ou un hyperplan qui minimise les erreurs quadratiques entre les valeurs cibles et les prédictions.Intuition et étapes : Le modèle prend en entrée une matrice de caractéristiques $X$ et un vecteur cible $y$. Il calcule une combinaison linéaire des entrées pour prédire la sortie, puis optimise ses coefficients par la méthode des moindres ordinaires pour minimiser l'erreur globale.Formule et symboles :$$\hat{y}_i = \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij}, \qquad J = \frac{1}{2n} \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$Définition des symboles : $n$ = nombre de lignes ; $p$ = nombre de caractéristiques ; $x_{ij}$ = variable $j$ de l'observation $i$ ; $y_i$ = vraie valeur cible ; $\beta_0$ = constante (biais) ; $\beta_j$ = coefficient estimé ; $J$ = coût moyen quadratique.Paramètres et hyperparamètres : Les paramètres appris sont les coefficients $\beta_0$ et $\beta_j$. L'hyperparamètre principal est fit_intercept (booléen pour inclure ou non la constante).Avantages et limites :Avantages : Modèle très rapide, hautement explicable et interprétable.Limites : Très sensible aux valeurs atypiques (outliers) et incapable de modéliser des relations non linéaires de base.Schéma explicatifDescription du schéma : Un nuage de points en deux dimensions représentant des observations $(x, y)$, traversé par une droite ajustée, avec des segments verticaux reliant chaque point à la droite.Comment lire ce schéma : Les segments verticaux matérialisent les résidus (les erreurs $y_i - \hat{y}_i$). L'algorithme cherche à positionner la droite de manière à minimiser la somme des carrés de ces segments verticaux.Contextes d'utilisationCas d'usage concrets :Finance / Gestion : Prévision du chiffre d'affaires d'une entreprise en fonction des dépenses publicitaires.   Logistique : Estimation du coût d'acheminement d'une marchandise selon son poids et la distance à parcourir.Nature des données et prérequis : Variables explicatives essentiellement numériques, relation sous-jacente linéaire, absence de colinéarité excessive entre les variables.Cas peu approprié : Données présentant des relations fortement non linéaires ou comportant de nombreux bruits et valeurs aberrantes.Métriques adaptées : MAE (Erreur Absolue Moyenne), RMSE (Racine de l'Erreur Quadratique Moyenne) et $R^2$ (pouvoir descriptif).   Analyse de l'expérience PythonEntrées : Les données d'entraînement Xr_ent et yr_ent issues d'un jeu de données synthétique (make_regression), séparées du jeu de test (Xr_test, yr_test).   Hyperparamètres : Utilisation des paramètres par défaut de LinearRegression() (fit_intercept=True).   Sortie obtenue : La valeur de la MAE calculée sur le jeu de test.   Conclusion : La sortie permet de quantifier l'erreur moyenne de prédiction sur le jeu de test synthétique. Cependant, elle ne permet pas d'établir de lien de causalité direct entre les variables, ni de garantir de la même performance sur des données économiques réelles.   Modèle type pour la Partie A : 8. Régression logistiquePrésentationNom et famille : Régression logistique / Logistic Regression, appartenant à la famille de la classification supervisée.Objectif : Transformer un score linéaire en une probabilité d'appartenance à une classe à l'aide d'une fonction sigmoïde.Intuition et étapes : Le modèle calcule une combinaison linéaire des caractéristiques (le logit), la convertit en une probabilité comprise entre 0 et 1 via la fonction logistique, puis applique un seuil de décision pour classifier l'observation.Formule et symboles :$$p(x) = \sigma(z) = \frac{1}{1 + e^{-z}}, \quad z = \beta_0 + \beta^\top x$$Définition des symboles : $p(x)$ = probabilité de la classe 1 ; $\sigma$ = fonction sigmoïde ; $z$ = score linéaire (logit) ; $\beta_0$ = biais ; $\beta$ = vecteur des poids ; $x$ = caractéristiques.Paramètres et hyperparamètres : Les coefficients $\beta_0$ et $\beta$ sont appris. L'hyperparamètre principal est $C$ (l'inverse de la force de régularisation).Avantages et limites :Avantages : Fournit des probabilités directement interprétables et un coût de calcul faible.Limites : Ne peut tracer que des frontières de décision linéaires dans l'espace d'origine, sauf si on lui fournit des transformations de variables.Schéma explicatifDescription du schéma : Une courbe en forme de "S" (la sigmoïde) tracée sur un graphique, avec un axe des abscisses représentant le score linéaire et un axe des ordonnées représentant la probabilité de 0 à 1, ainsi qu'une ligne de seuil à 0,5.Comment lire ce schéma : Tout score supérieur au seuil (généralement 0,5) positionne l'observation du côté de la classe positive (1), tandis qu'un score inférieur la classe du côté de la classe négative (0).Contextes d'utilisationCas d'usage concrets :Finance : Prédiction du défaut de paiement d'un client (scoring de crédit).   Marketing : Prédire si un client va souscrire ou non à une offre promotionnelle après une campagne.Nature des données et prérequis : Cible binaire (0 ou 1), caractéristiques numériques idéalement standardisées pour stabiliser l'optimisation.Cas peu approprié : Problèmes de classification dont les frontières entre classes sont extrêmement complexes et non linéaires.Métriques adaptées : F1-score, Précision, Rappel, ROC-AUC et PR-AUC (particulièrement si les classes sont déséquilibrées). Hyperparamètres : max_iter=1000 et standardisation des variables via un Pipeline pour éviter la fuite de données.   Sortie obtenue : Le score F1 calculé sur le jeu de test.
```python
# Données pédagogiques reproductibles (les étiquettes sont connues pour l'évaluation)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression, make_classification, make_blobs, load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score, accuracy_score, f1_score, silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
GRAINE = 42
X_reg, y_reg = make_regression(n_samples=280, n_features=5, noise=13, random_state=GRAINE)
X_cls, y_cls = make_classification(n_samples=400, n_features=6, n_informative=4, n_redundant=0, random_state=GRAINE)
X_clu, etiquettes_reference = make_blobs(n_samples=240, centers=3, cluster_std=0.85, random_state=GRAINE)
Xr_ent, Xr_test, yr_ent, yr_test = train_test_split(X_reg, y_reg, test_size=.25, random_state=GRAINE)
Xc_ent, Xc_test, yc_ent, yc_test = train_test_split(X_cls, y_cls, test_size=.25, stratify=y_cls, random_state=GRAINE)
print('Régression:',X_reg.shape,'Classification:',X_cls.shape)
```

## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# Définition de la fonction de rapport manquante
def rapport_reg(modele, titre):
    predictions = modele.predict(Xr_test)
    mae = mean_absolute_error(yr_test, predictions)
    r2 = r2_score(yr_test, predictions)
    print(f"=== {titre} ===")
    print(f"R²: {r2:.3f} | MAE: {mae:.3f}")

# Entraînement et évaluation
modele = LinearRegression().fit(Xr_ent, yr_ent)
rapport_reg(modele, '1. Régression linéaire')
print('Constante:', round(float(modele.intercept_), 3), '| Coefficients:', np.round(modele.coef_, 2))
```

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Ridge

# Modèle Ridge avec alpha=1.0
modele = make_pipeline(StandardScaler(), Ridge(alpha=1.0)).fit(Xr_ent, yr_ent)
rapport_reg(modele, '2. Ridge (alpha=1.0)')
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
modele = make_pipeline(StandardScaler(), Lasso(alpha=.1, max_iter=10000)).fit(Xr_ent, yr_ent)
rapport_reg(modele, '3. Lasso (alpha=0.1)')
print('Coefficients nuls (alpha=0.1):', int((modele[-1].coef_ == 0).sum()), 'sur', X_reg.shape[1])
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

# Pipeline Polynomiale (degré 2) + Ridge
modele = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(degree=2, include_bias=False),
    Ridge(alpha=10)
).fit(Xr_ent, yr_ent)

# Rapport des performances
rapport_reg(modele, '4. Polynomiale (degré 2) + Ridge')
print('Nombre de variables après expansion (degré 2):', modele[1].n_output_features_)
```

```python
from sklearn.tree import DecisionTreeRegressor

# Modèle Arbre de décision (max_depth=5)
modele = DecisionTreeRegressor(max_depth=5, random_state=GRAINE).fit(Xr_ent, yr_ent)

# Rapport des performances
rapport_reg(modele, '5. Arbre de régression (max_depth=5)')
```

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import numpy as np

generateur = np.random.default_rng(GRAINE)
vraies_probas = np.array([.2, .5, .7])
somme = np.zeros(3)
comptes = np.zeros(3)

# Démonstration simple initiale
for t in range(500):
    bras = t % 3
    gain = int(generateur.random() < vraies_probas[bras])
    somme[bras] += gain
    comptes[bras] += 1
print('Valeurs estimées (démo initiale) :', somme / comptes)

# Définition propre de la fonction pour l'évaluation sur 20 graines
def bandit_uniforme(graine):
    g = np.random.default_rng(graine)
    somme_loc = np.zeros(3)
    comptes_loc = np.zeros(3)
    cumul = 0
    T = 500
    for t in range(T):
        bras = t % 3
        gain = int(g.random() < vraies_probas[bras])
        somme_loc[bras] += gain
        comptes_loc[bras] += 1
        cumul += gain
    return somme_loc / comptes_loc, cumul

estimees, cumul = bandit_uniforme(GRAINE)
print('=== 25. Bandits manchots ===')
print('Valeurs estimées (graine 42) :', np.round(estimees, 3), '| Récompense cumulée :', cumul)
print('Récompense cumulée espérée du meilleur bras seul (500 × 0,7) :', int(500 * vraies_probas.max()))

cumuls = [bandit_uniforme(s)[1] for s in range(20)]
print('Allocation uniforme sur 20 graines : moyenne =', round(float(np.mean(cumuls)), 1), '| écart-type =', round(float(np.std(cumuls)), 1))
```

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```
